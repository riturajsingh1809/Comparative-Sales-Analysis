```python
# Import Libraries

import pandas as pd
import numpy as np


# Load Dataset

df = pd.read_excel("Cleaned_Dataset.xlsx")


# Display First 5 Rows

df.head()


# Check Number of Rows and Columns

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])


# Check Column Names

print(df.columns.tolist())


# Check Dataset Information

df.info()


# Check Missing Values

print("\nMissing Values:")
print(df.isnull().sum())


# Check Duplicate Values

print("\nDuplicate Rows:", df.duplicated().sum())


# Statistical Summary

df.describe()


# Overall Sales Analysis

total_sales = df["sales"].sum()

print("Total Sales:", total_sales)


# Overall Profit Analysis

total_profit = df["profit"].sum()

print("Total Profit:", total_profit)


# Total Units

total_units = df["units"].sum()

print("Total Units:", total_units)


# Total Records

total_records = len(df)

print("Total Records:", total_records)


# Average Sales

average_sales = df["sales"].mean()

print("Average Sales:", average_sales)


# Region-wise Analysis

region_analysis = df.groupby("region").agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Units=("units", "sum"),
    Total_Records=("customer_id", "count"),
    Average_Sales=("sales", "mean")
).reset_index()

region_analysis = region_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

print("\nRegion-wise Analysis:")
display(region_analysis)


# Plan-wise Analysis

plan_analysis = df.groupby("plan").agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Units=("units", "sum"),
    Total_Records=("customer_id", "count"),
    Average_Sales=("sales", "mean")
).reset_index()

plan_analysis = plan_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

print("\nPlan-wise Analysis:")
display(plan_analysis)


# Channel-wise Analysis

channel_analysis = df.groupby("channel").agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Units=("units", "sum"),
    Total_Records=("customer_id", "count"),
    Average_Sales=("sales", "mean")
).reset_index()

channel_analysis = channel_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

print("\nChannel-wise Analysis:")
display(channel_analysis)


# Monthly Analysis

monthly_analysis = df.groupby(
    ["month", "month_name"]
).agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Units=("units", "sum"),
    Total_Records=("customer_id", "count")
).reset_index()

monthly_analysis = monthly_analysis.sort_values(
    by="month"
)

print("\nMonthly Analysis:")
display(monthly_analysis)


# Region and Plan Analysis

region_plan_analysis = df.groupby(
    ["region", "plan"]
).agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Units=("units", "sum"),
    Total_Records=("customer_id", "count")
).reset_index()

region_plan_analysis = region_plan_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

print("\nRegion and Plan Analysis:")
display(region_plan_analysis)


# Region and Channel Analysis

region_channel_analysis = df.groupby(
    ["region", "channel"]
).agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Units=("units", "sum"),
    Total_Records=("customer_id", "count")
).reset_index()

region_channel_analysis = region_channel_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

print("\nRegion and Channel Analysis:")
display(region_channel_analysis)


# Find Top Region

top_region = region_analysis.iloc[0]["region"]

print("Top Region:", top_region)


# Find Lowest Region

lowest_region = region_analysis.iloc[-1]["region"]

print("Lowest Region:", lowest_region)


# Find Top Plan

top_plan = plan_analysis.iloc[0]["plan"]

print("Top Plan:", top_plan)


# Find Top Channel

top_channel = channel_analysis.iloc[0]["channel"]

print("Top Channel:", top_channel)


# Find Highest Sales Month

highest_sales_month = monthly_analysis.loc[
    monthly_analysis["Total_Sales"].idxmax(),
    "month_name"
]

print("Highest Sales Month:", highest_sales_month)


# Find Lowest Sales Month

lowest_sales_month = monthly_analysis.loc[
    monthly_analysis["Total_Sales"].idxmin(),
    "month_name"
]

print("Lowest Sales Month:", lowest_sales_month)


# Create Overall Summary

overall_analysis = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Units",
        "Total Records",
        "Average Sales",
        "Top Region",
        "Lowest Region",
        "Top Plan",
        "Top Channel",
        "Highest Sales Month",
        "Lowest Sales Month"
    ],

    "Value": [
        total_sales,
        total_profit,
        total_units,
        total_records,
        average_sales,
        top_region,
        lowest_region,
        top_plan,
        top_channel,
        highest_sales_month,
        lowest_sales_month
    ]
})

print("\nOverall Summary:")
display(overall_analysis)


# Save All Analysis Results in One Excel File

with pd.ExcelWriter(
    "Analysis_Result.xlsx",
    engine="openpyxl"
) as writer:

    overall_analysis.to_excel(
        writer,
        sheet_name="Overall Summary",
        index=False
    )

    region_analysis.to_excel(
        writer,
        sheet_name="Region Analysis",
        index=False
    )

    plan_analysis.to_excel(
        writer,
        sheet_name="Plan Analysis",
        index=False
    )

    channel_analysis.to_excel(
        writer,
        sheet_name="Channel Analysis",
        index=False
    )

    monthly_analysis.to_excel(
        writer,
        sheet_name="Monthly Analysis",
        index=False
    )

    region_plan_analysis.to_excel(
        writer,
        sheet_name="Region Plan",
        index=False
    )

    region_channel_analysis.to_excel(
        writer,
        sheet_name="Region Channel",
        index=False
    )


print("\nAnalysis completed successfully!")
print("Output file: Analysis_Result.xlsx")
```
